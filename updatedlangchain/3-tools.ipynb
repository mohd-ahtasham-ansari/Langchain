### Tools

Models can request to call tools that perform tasks such as fetching data from a database , searching and running code,
Tools are pairing of:


  1.A Schema,inclunding the name of the tools, a description , and/or argument definition (often a JSON schema)


  2.A function of coroutine to execute

```python
import os
from dotenv import find_dotenv,load_dotenv

load_dotenv(find_dotenv())

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

from langchain_groq import ChatGroq

model=ChatGroq(model="llama-3.1-8b-instant")
response=model.invoke("what is groq")
response.content
```
