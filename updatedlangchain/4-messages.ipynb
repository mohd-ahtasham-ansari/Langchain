### Message prompt

Alternatively, you can pass in a list of messages to the model by providing a list of message objects.

Message types

* System message - Tells the model how to behave and provide context for interactions
* Human message - Represents user input and interactions with the model
* AI message - Responses generated by the model, including text content, tool calls, and metadata
* Tool message - Represents the outputs of tool calls

### System Message

A SystemMessage represent an initial set of instructions that primes the model's behavior. You can use a system message to set the tone, define the model's role, and establish guidelines for responses.

### Human Message

Human message represent the user input and interactions .They can contain text , images, audio ,files and other amount of multimodel content

### AI message

AI message represent the output of model invocation. They can contain multimodel , text , audio , video and other amount of data

### Tool Message

for models that support tool calling, AI message can contain tool call .  Tool message are used to pass the results of a single tool execution back to the model

```python
from langchain.messages import AIMessage,HumanMessage,ToolCall,SystemMessage
```

```python
message=[
    SystemMessage("you are a poetry expert"),
    HumanMessage("write a poetry on placement preparation ")
]

response = model.invoke(message)
response.content
```

```python
system_msg=SystemMessage("You are a expert coding assistant")
message=[
    system_msg,
    HumanMessage("how to create a RESTapi?")
]
for chunk in model.stream(message):
    print(chunk.text , end="", flush=True)
```

```python
from langchain.messages import AIMessage
from langchain.messages import ToolMessage
from langchain.messages import HumanMessage

# after a model makes a tool call
# here we domenstrated manually creating the message for brevity
ai_msg= AIMessage(
    content=[],
    tool_calls=[{
        "name":"get_weather",
        "args":{"location":"san francisco"},
        "id":"abc123"
    }]
)
weather_result=" sunny , 72F"
tool_message=ToolMessage(
    content=weather_result,
    tool_call_id="abc123" # must match call id
)
message=[
    HumanMessage("how is the weather in san francisco?"),
    ai_msg,
    tool_message
]
response = model.invoke(message)
```
